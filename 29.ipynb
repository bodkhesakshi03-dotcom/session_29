```python
import pandas as pd
from sklearn.model_selection import train_test_split

url = "https://raw.githubusercontent.com/PreetiSharma2815/Emotions-Dataset/main/train.txt"
df = pd.read_csv(url, sep=";", names=["text", "emotion"])

#  SEPARATE THE FEATURES
X = df['text']
y = df['emotion']

# PERFORM TRAIN-TEST SPLIT
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Print the shape of the splits
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Initialize CountVectorizer
bow_vectorizer = CountVectorizer()

# Fit and transform X_train
X_train_bow = bow_vectorizer.fit_transform(X_train)

# Transform X_test
X_test_bow = bow_vectorizer.transform(X_test)

# Print the shape of the resulting matrices
print("X_train_bow shape:", X_train_bow.shape)
print("X_test_bow shape:", X_test_bow.shape)

# Display the first 20 feature names
print("First 20 vocabulary words:", bow_vectorizer.get_feature_names_out()[:20])
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Initialize the model
mnb_bow = MultinomialNB()

# Fit the model on X_train_bow and y_train
mnb_bow.fit(X_train_bow, y_train)

# Make predictions on the test set
y_pred_bow = mnb_bow.predict(X_test_bow)

# Calculate and print the accuracy score
accuracy_bow = accuracy_score(y_test, y_pred_bow)
print(f"Bag of Words + MultinomialNB Accuracy: {accuracy_bow:.4f}")
```

```python
import random

# Print the total size of the vocabulary
vocab_names = bow_vectorizer.get_feature_names_out()
print("Total size of the vocabulary:", len(vocab_names))

# Show any 15 words from the vocabulary
print("15 random words from vocabulary:", random.sample(list(vocab_names), 15))

# Convert one sample document from the training set into its BoW vector
sample_doc = X_train.iloc[0]
sample_vector = bow_vectorizer.transform([sample_doc])

print("\nSample Document Text:\n", sample_doc)
print("\nSample Document Vector representation (Sparse):\n", sample_vector)
print("\nSample Document Vector representation (Dense Array):\n", sample_vector.toarray())
```

```python
# Create a new CountVectorizer that uses bigrams
ngram_vectorizer = CountVectorizer(ngram_range=(1, 2))

# Fit and transform the training data
X_train_ngram = ngram_vectorizer.fit_transform(X_train)

# Transform the test data
X_test_ngram = ngram_vectorizer.transform(X_test)

# Print the shape of the new feature matrix
print("X_train_ngram shape:", X_train_ngram.shape)

# Display some of the bigram feature
ngram_features = ngram_vectorizer.get_feature_names_out()
bigram_only_features = [f for f in ngram_features if ' ' in f]
print("Sample Bigram Features:", bigram_only_features[:10])
```

```python
# Train a MultinomialNB model using the bigram features
mnb_ngram = MultinomialNB()
mnb_ngram.fit(X_train_ngram, y_train)

# Predictions and accuracy
y_pred_ngram = mnb_ngram.predict(X_test_ngram)
accuracy_ngram = accuracy_score(y_test, y_pred_ngram)

# Calculate and print the accuracy on the test set
print(f"Bigram + MultinomialNB Accuracy: {accuracy_ngram:.4f}")

# Compare this accuracy with the unigram (basic BoW) accuracy
print(f"Difference in accuracy (Bigram - Unigram): {(accuracy_ngram - accuracy_bow):.4f}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Apply TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

# Fit and transform X_train
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)

# Transform X_test
X_test_tfidf = tfidf_vectorizer.transform(X_test)

# Print the shape of the TF-IDF matrices
print("X_train_tfidf shape:", X_train_tfidf.shape)
print("X_test_tfidf shape:", X_test_tfidf.shape)

# Display the first 15 feature names
print("First 15 TF-IDF features:", tfidf_vectorizer.get_feature_names_out()[:15])
```

```python
# Train a Multinomial Naive Bayes model using the TF-IDF features
mnb_tfidf = MultinomialNB()

# Fit the model
mnb_tfidf.fit(X_train_tfidf, y_train)

# Make predictions on the test set
y_pred_tfidf = mnb_tfidf.predict(X_test_tfidf)

# Calculate and print the accuracy score
accuracy_tfidf = accuracy_score(y_test, y_pred_tfidf)
print(f"TF-IDF + MultinomialNB Accuracy: {accuracy_tfidf:.4f}")
```

```python
import pandas as pd

# Create a comparison table of the three approaches
results = {
    "Vectorization Approach": [
        "Bag of Words (Unigrams)",
        "Bag of Words (Unigrams + Bigrams)",
        "TF-IDF"
    ],
    "Accuracy Score": [
        round(accuracy_bow, 4),
        round(accuracy_ngram, 4),
        round(accuracy_tfidf, 4)
    ]
}

comparison_df = pd.DataFrame(results)
print(comparison_df.to_string(index=False))
```

```python
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# 1. Load and clean the text
url = "https://raw.githubusercontent.com/PreetiSharma2815/Emotions-Dataset/main/train.txt"
df = pd.read_csv(url, sep=";", names=["text", "emotion"])

# 2. Perform train-test split
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['emotion'], test_size=0.20, random_state=42
)

# 3. Apply both CountVectorizer and TfidfVectorizer
count_vec = CountVectorizer()
X_train_c = count_vec.fit_transform(X_train)
X_test_c = count_vec.transform(X_test)

tfidf_vec = TfidfVectorizer()
X_train_t = tfidf_vec.fit_transform(X_train)
X_test_t = tfidf_vec.transform(X_test)

# 4. Train MultinomialNB on both
model_c = MultinomialNB()
model_c.fit(X_train_c, y_train)

model_t = MultinomialNB()
model_t.fit(X_train_t, y_train)

# 5. Compare their accuracy
acc_c = accuracy_score(y_test, model_c.predict(X_test_c))
acc_t = accuracy_score(y_test, model_t.predict(X_test_t))

print(f"CountVectorizer Pipeline Accuracy: {acc_c:.4f}")
print(f"TfidfVectorizer Pipeline Accuracy: {acc_t:.4f}")

# 6. Save the best vectorizer and the best model using joblib
if acc_c >= acc_t:
    print("Saving CountVectorizer and its Model...")
    joblib.dump(count_vec, 'best_vectorizer.pkl')
    joblib.dump(model_c, 'best_model.pkl')
else:
    print("Saving TfidfVectorizer and its Model...")
    joblib.dump(tfidf_vec, 'best_vectorizer.pkl')
    joblib.dump(model_t, 'best_model.pkl')

print("Assets saved successfully.")
```
